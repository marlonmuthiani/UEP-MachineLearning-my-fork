Importing packages

```python
from sklearn.metrics import (
    classification_report, confusion_matrix, recall_score,
    precision_score, f1_score, roc_auc_score,
    mean_squared_error, mean_absolute_error, r2_score
)
from sklearn.metrics import roc_curve, precision_recall_curve
import matplotlib.pyplot as plt
import pandas as pd
```

Klasyfikator

```python
df = pd.read_csv("path/to/your/data.csv")
X = df.drop(columns="volume_total")
y = (df["volume_total"] > 10_000).astype(int)
```

```python
best_model = joblib.load("")
```

```python
y_pred = best_model.predict(X)
y_proba = best_model.predict_proba(X)[:, 1]
```

```python
print("=== Classification Report ===")
print(classification_report(y, y_pred, target_names=["<=10k", ">10k"]))

print("\n=== Confusion Matrix ===")
cm = confusion_matrix(y, y_pred)
print(cm)
print(f"TN: {cm[0,0]}, FP: {cm[0,1]}")
print(f"FN: {cm[1,0]}, TP: {cm[1,1]}")

recall = recall_score(y, y_pred)
precision = precision_score(y, y_pred)
f1 = f1_score(y, y_pred)
auc = roc_auc_score(y, y_proba)

print(f"\nRecall: {recall:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1: {f1:.4f}")
print(f"AUC: {auc:.4f}")

fnr = cm[1,0] / (cm[1,0] + cm[1,1]) if (cm[1,0] + cm[1,1]) > 0 else 0
print(f"False Negative Rate: {fnr:.4f}")
```

```python
fpr, tpr, _ = roc_curve(y, y_proba)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(fpr, tpr, label=f"AUC={auc:.3f}")
plt.plot([0,1], [0,1], "k--", label="Random")
plt.xlabel("FPR")
plt.ylabel("TPR (Recall)")
plt.title("ROC Curve")
plt.legend()
plt.grid()

prec, rec, _ = precision_recall_curve(y, y_proba)
plt.subplot(1, 2, 2)
plt.plot(rec, prec)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.grid()

plt.tight_layout()
plt.show()
```
