```python
df = pd.read_csv("../data_model/db_additional_fe.csv")
df_c = pd.read_csv("../data_model/db_collected_fe.csv")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split


best_regressor = RandomForestRegressor(n_estimators=442, max_depth=14,min_samples_split=8,min_samples_leaf=2,max_features="sqrt")
X = df.drop(columns='volume_total')
X = X.filter(regex="emb")
y = df['volume_total']
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2) 
best_regressor.fit(X_train,y_train)
y_pred = best_regressor.predict(X_test)
rmse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.4f}")
print(f"MAE: {mae:.4f}")
print(f"R²: {r2:.4f}")
```
